```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder #convert categorical columns to numerical columns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

# Data Exploration
* To understand what the data is about.

```python
#load the data
ds=pd.read_csv("Pokemon Data.csv")
ds.head()
```

```python
#dataset information
ds.info()
```

```python
ds.isnull().sum()
```

```python
#check for duplicates
ds.duplicated().sum()
```

# Data Preprocessing
* Prepare the data for analysis.

```python
#dropping null values
ds.dropna(inplace=True)
```

```python
#convert categorical columns to numerical
ls=LabelEncoder()
ds["Type 1"]=ls.fit_transform(ds["Type 1"])
ds["Type 2"]=ls.fit_transform(ds["Type 2"])
ds["Legendary"]=ls.fit_transform(ds["Legendary"])
```

```python
#drop unecessary/unique columns
ds.drop('#',axis=1,inplace=True)
```

```python
ds.head()
```

```python

#feature extraction
x=ds.drop('Legendary',axis=1)
y=ds['Legendary']
```

```python
#split data into testing and training
x_train,x_test,y_train,y_test=train_test_split(x,y, test_size=0.2)
```

# Model Selection
* Load different models and check their accuracy to check the best model to make predictions on the data.

#1. Logistic Regression

```python
#load the model
log_model=LogisticRegression()

#train the model
log_model.fit(x_train,y_train)

#make predictions
predictions=log_model.predict(x_test)

#evaluate the model
accuracy_score(y_test,predictions)
```

# Findings
* Logistic regression accuracy is 100%

#2. KNN

```python
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train,y_train)
predictions=knn.predict(x_test)
accuracy_score(y_test,predictions)
```

# Findings
* KNN accuracy is 94%

#3. Decision Tree

```python
dt=DecisionTreeClassifier()
dt.fit(x_train,y_train)
predictions=dt.predict(x_test)
accuracy_score(y_test,predictions)
```

# Findings
* Decision tree has an accuracy of 100%

#ALGORITHM RECOMMENDATION
* Logistic regresion and decision Tree are the best algorithms to make predictions on whether a pokemon is Legendary or not.
